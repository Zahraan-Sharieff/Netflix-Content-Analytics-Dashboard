# 🍿 Content Distribution Analysis

This visualization compares the distribution of Movies and TV Shows available on Netflix.

```python
type_counts = df['type'].value_counts()

fig = go.Figure(data=[go.Pie(
    labels=type_counts.index,
    values=type_counts.values,
    hole=0.55,
    marker=dict(
        colors=[NETFLIX_RED, '#222222'],
        line=dict(color='black', width=4)
    ),
    textfont=dict(color='white', size=18)
)])

fig.update_layout(
    title={
        'text': 'Netflix Content Distribution',
        'x': 0.5,
        'font': {'size': 28, 'color': 'white'}
    },
    paper_bgcolor='black',
    plot_bgcolor='black',
    font=dict(color='white'),
    showlegend=True,
    legend=dict(
        font=dict(size=14),
        bgcolor='black'
    )
)

fig.show()
```

# 🌍 Top Countries Producing Netflix Content

This visualization highlights the top countries contributing the highest number of movies and TV shows on Netflix.

```python
country_data = df['country'].dropna().str.split(', ').explode()

top_countries = country_data.value_counts().head(10)

fig = px.bar(
    x=top_countries.values,
    y=top_countries.index,
    orientation='h',
    text=top_countries.values,
    color=top_countries.values,
    color_continuous_scale=['#330000', '#E50914']
)

fig.update_layout(
    title='Top 10 Countries Producing Netflix Content',
    title_x=0.5,
    paper_bgcolor='black',
    plot_bgcolor='black',
    font=dict(color='white', size=14),
    xaxis_title='Number of Titles',
    yaxis_title='Country',
    coloraxis_showscale=False
)

fig.update_traces(
    textposition='outside'
)

fig.show()
```

# 📈 Netflix Content Growth Over Years

This visualization shows how Netflix content has grown over time based on the release year of movies and TV shows available on the platform.

```python
release_year = df['release_year'].value_counts().sort_index()

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=release_year.index,
    y=release_year.values,
    mode='lines+markers',
    line=dict(color=NETFLIX_RED, width=4),
    marker=dict(size=7),
    fill='tozeroy',
    fillcolor='rgba(229, 9, 20, 0.25)'
))

fig.update_layout(
    title='Netflix Content Growth Over Years',
    title_x=0.5,
    paper_bgcolor='black',
    plot_bgcolor='black',
    font=dict(color='white'),
    xaxis=dict(
        title='Release Year',
        showgrid=False
    ),
    yaxis=dict(
        title='Number of Shows/Movies',
        showgrid=False
    )
)

fig.show()
```
